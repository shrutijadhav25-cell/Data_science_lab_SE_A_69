```python
import heapq

def a_star_search(graph, start, goal, heuristic):

    # open_set is a priority queue of (f_score, node)
    open_set = [(0 + heuristic[start], start)]

    # came_from maps current node to previous node
    came_from = {}

    # g_score is the cost from start to current node
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    # f_score is g_score + heuristic for current node
    f_score = {node: float('inf') for node in graph}
    f_score[start] = heuristic[start]

    while open_set:
        # Get the node with the lowest f_score
        current_f_score, current_node = heapq.heappop(open_set)

        if current_node == goal:
            # Reconstruct path
            path = []
            while current_node in came_from:
                path.append(current_node)
                current_node = came_from[current_node]
            path.append(start)
            return path[::-1], g_score[goal]

        for neighbor, cost in graph[current_node].items():
            tentative_g_score = g_score[current_node] + cost

            if tentative_g_score < g_score[neighbor]:
                came_from[neighbor] = current_node
                g_score[neighbor] = tentative_g_score
                f_score[neighbor] = tentative_g_score + heuristic[neighbor]
                heapq.heappush(open_set, (f_score[neighbor], neighbor))

    return None, float('inf') # No path found
```

```python
# Example Usage:

# Define the graph (adjacency list with costs)
# For example, a road network where nodes are cities and edges are roads with distances
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'D': 2, 'E': 5},
    'C': {'A': 4, 'F': 2},
    'D': {'B': 2, 'G': 1},
    'E': {'B': 5, 'G': 1},
    'F': {'C': 2, 'G': 3},
    'G': {}
}

# Define the heuristic function (estimated distance from each node to the goal 'G')
# This is typically an admissible heuristic (never overestimates the true cost)
heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 1,
    'F': 3,
    'G': 0  # Heuristic for the goal node is always 0
}

start_node = 'A'
goal_node = 'G'

path, cost = a_star_search(graph, start_node, goal_node, heuristic)

if path:
    print(f"Shortest path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}")
```
