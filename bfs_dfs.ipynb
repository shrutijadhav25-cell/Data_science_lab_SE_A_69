```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("Graph:", graph)
```

```python
from collections import deque

def bfs(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)

    bfs_traversal_order = []

    while queue:
        current_node = queue.popleft()
        bfs_traversal_order.append(current_node)


        for neighbor in graph[current_node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return bfs_traversal_order

print("BFS Traversal starting from 'A':", bfs(graph, 'A'))
```

```python
def dfs(graph, start_node):
    visited = set()
    stack = [start_node]

    dfs_traversal_order = []

    while stack:
        current_node = stack.pop()

        if current_node not in visited:
            visited.add(current_node)
            dfs_traversal_order.append(current_node)


            for neighbor in reversed(graph[current_node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return dfs_traversal_order

print("DFS Traversal starting from 'A':", dfs(graph, 'A'))
```
