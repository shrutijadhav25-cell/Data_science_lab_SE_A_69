```python
# Coefficients
a1, b1, c1 = 2, 3, 10
a2, b2, c2 = 4, 5, 20

# Make the x coefficients the same
m1 = a2
m2 = a1

# Multiply equations to match x terms
A1 = a1 * m1
B1 = b1 * m1
C1 = c1 * m1

A2 = a2 * m2
B2 = b2 * m2
C2 = c2 * m2

# Subtract equations to eliminate x
y = (C1 - C2) / (B1 - B2)

# Substitute y back to find x
x = (c1 - b1 * y) / a1

print("x =", x)
print("y =", y)
```
