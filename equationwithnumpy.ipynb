```python
import numpy as np

# Coefficient matrix
A = np.array([
    [2, 3],
    [4, 5]
])

# Constants
B = np.array([10, 20])

# Solve Ax = B
solution = np.linalg.solve(A, B)

x = solution[0]
y = solution[1]

print("x =", x)
print("y =", y)
```
