```python
import pandas as pd

# Load customer personality dataset
df = pd.read_csv("/content/archive.zip", sep="\t")

# ------------------ FILTERING ------------------
# Example: customers with income greater than 50k and recency less than 100
filtered = df[(df["Income"] > 50000) & (df["Recency"] < 100)]

print("Filtered dataset (high income & recent customers):")
print(filtered.head())

# ------------------ GROUPING + AGGREGATION ------------------
grouped = df.groupby("Marital_Status").agg(
    Total_Income=("Income", "sum"),
    Mean_Income=("Income", "mean"),
    Count_Customers=("Income", "count"),
    Min_Income=("Income", "min"),
    Max_Income=("Income", "max")
)

print("\nAggregated statistics by Marital Status:")
print(grouped)
```
