```python
# Minimax Algorithm with Game Tree (Graph Representation)

# Minimax function
def minimax(node, depth, is_max):
    # If leaf node
    if depth == 2:
        return node

    if is_max:
        best = -1000
        for child in node:
            val = minimax(child, depth + 1, False)
            best = max(best, val)
        return best
    else:
        best = 1000
        for child in node:
            val = minimax(child, depth + 1, True)
            best = min(best, val)
        return best


# Game tree (graph)
# MAX
# ├── MIN → [3, 5]
# └── MIN → [2, 9]

game_tree = [
    [3, 5],
    [2, 9]
]

# Printing the graph
print("Game Tree (Graph):")
print(" MAX")
print(" / \\")
print(" MIN MIN")
print(" / \\ / \\")
print(" 3 5 2 9")

# Performing Minimax
result = minimax(game_tree, 0, True)

print("\nOptimal value using Minimax:", result)
```
