```python
data_input = input("Enter numbers separated by spaces: ")
numbers_str = data_input.split()

dataset = []
for s in numbers_str:
    try:
        dataset.append(float(s))
    except ValueError:
        print(f"Skipping invalid input: {s}")

if not dataset:
    print("No valid numbers entered. Exiting.")
else:
    print(f"Dataset: {dataset}")
```

```python
if dataset:
    data_sum = 0
    data_count = 0
    for num in dataset:
        data_sum += num
        data_count += 1

    mean = data_sum / data_count
    print(f"Mean: {mean}")
```

```python
if dataset:
    # Create a sorted copy of the dataset
    sorted_dataset = list(dataset)
    n = len(sorted_dataset)

    # Manual bubble sort for demonstration (can use sorted_dataset.sort() in practice)
    for i in range(n):
        for j in range(0, n-i-1):
            if sorted_dataset[j] > sorted_dataset[j+1]:
                sorted_dataset[j], sorted_dataset[j+1] = sorted_dataset[j+1], sorted_dataset[j]

    if n % 2 == 1:
        median = sorted_dataset[n // 2]
    else:
        mid1 = sorted_dataset[n // 2 - 1]
        mid2 = sorted_dataset[n // 2]
        median = (mid1 + mid2) / 2

    print(f"Sorted Dataset: {sorted_dataset}")
    print(f"Median: {median}")
```

```python
if dataset:
    squared_differences_sum = 0
    for num in dataset:
        squared_differences_sum += (num - mean) * (num - mean)

    variance = squared_differences_sum / data_count
    print(f"Variance: {variance}")
```

```python
if dataset:
    # Simple iterative square root function
    def custom_sqrt(n, iterations=100):
        if n < 0:
            raise ValueError("Cannot calculate square root of a negative number.")
        if n == 0:
            return 0.0
        x = n # Initial guess
        for _ in range(iterations):
            x = 0.5 * (x + n / x)
        return x

    standard_deviation = custom_sqrt(variance)
    print(f"Standard Deviation: {standard_deviation}")
```
