```python
import numpy as np

# Input 3x3 matrix
A = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

det = np.linalg.det(A)

if det == 0:
    print("Matrix is not invertible (determinant is 0).")
else:
    A_inv = np.linalg.inv(A)
    print("Inverse of the matrix:")
    print(A_inv)
```
